# Encoding & Embedding

- **Encoding:** Numerical representation of data.
- **Embedding:** Meaningful numerical representation that captures semantic meaning.

## What is Encoding?

Encoding is the process of converting raw data (text) into numerical form (often sparse) without necessarily preserving semantic meaning.

### Types of Encoding

1. **One-Hot Encoding**
2. **BoW (Bag of Words)**
3. **TF-IDF (Term Frequency–Inverse Document Frequency)**
4. **N-grams**

### BM25

BM25 is an improved extension of TF-IDF, but technically, it is a ranking algorithm, not just an encoding method.

### GloVe

GloVe (Global Vectors) is based on word co-occurrence statistics(count-based) and matrix factorization.

# Vector

A vector is a set of numerical values.

- `5` → Scalar (not a vector)
- `[5, 10]` → Vector

## Vectors and Matrices

- **Matrix:** Collection of vectors.
- **Vector:** A special case of a matrix (one row or one column).
- **Matrix:** A table of numbers (grid).

A vector is a numerical representation of data in n-dimensional space.

- **n-dimensional vector:** `[1, 2, 3, 4, 5, 6, ..., n]`

In mathematics, a vector combines magnitude and direction.

- A vector = List of numbers `[1.2, -0.5, 3.8]`
- In NLP, embedding = Vector with meaning.
- Captures semantic relationships.

## Dot Product

The **dot product** combines corresponding components of two vectors and returns a single number (a scalar).

For 2D vectors \(A=(x_1,y_1)\) and \(B=(x_2,y_2)\):

\[
A \cdot B = x_1x_2 + y_1y_2
\]

### Example: Perpendicular Unit Vectors

Let:

\[
A = (1,0), \qquad B = (0,1)
\]

```text
        ↑ B = (0, 1)
        |
        |
        O────→ A = (1, 0)
```

Calculate the dot product by multiplying matching components and adding the results:

\[
A \cdot B = (1 \times 0) + (0 \times 1)
\]

\[
A \cdot B = 0 + 0 = 0
\]

The result is **0** because the vectors are perpendicular (at a 90° angle).

## What is Embedding?

Embedding is a mathematical representation of data.

Embedding converts text, images, audio, and video into meaningful numbers.

Embedding is a dense, high-dimensional numerical vector that represents the semantic meaning of data.

## Embeddings

- **Dimensions:** Typically range from 100 to several thousand dimensions, depending on the model (e.g., 348, 768, 1024, 1536, 2048).

- **Vector:** A mathematical object in linear algebra. In programming, a vector is typically stored as an array, meaning a collection of numerical values.

- **Convert:** Data into numerical vectors that preserve meaning.

### Example

- Embedding("king") and Embedding("queen") → Small angle between them (less distance).
- Embedding("king") and Embedding("banana") → Larger angle between them (greater distance).

### Distance / Similarity Measures

- Cosine Similarity
- Dot Product
- Euclidean Distance (ED)

### Classical Word Embedding Models

- Word2Vec
- GloVe
- FastText

### Limitations

- They create one embedding for a word regardless of context.
- They produce static embeddings (fixed vectors).
- They do not understand context.

## Example: Static vs Dynamic Embeddings

**Sentence 1:** I sat at the river bank.  
**Sentence 2:** I deposited money in the bank.

**Word:** bank

In a static embedding model (Word2Vec), the word "bank" is represented by the same vector in both sentences.

```text
bank → [0.21,-0.44,0.89,....]
```

In SOTA Transformer-based embedding models, the word "bank" is represented by different vectors depending on the context.

\[
\text{bank} \rightarrow [0.21, -0.44, 0.89, \ldots]
\]

\[
\text{bank} \rightarrow [0.93, 0.001, 0.75, \ldots]
\]

## Embedding

Data (text, image, video, audio) → Vector

- **Classical:** Neural Networks → Word2Vec → Static embeddings
- **SOTA:** Transformers → Dynamic embeddings (e.g., BERT, OpenAI embedding models)

## Difference Between Encoding and Embedding

| Feature | Encoding (Frequency) | Embedding |
|---|---|---|
| Purpose | Convert data into numbers | Convert data into numbers representing meaning |
| Nature | Often sparse (BoW) | Dense |
| Semantic meaning | Not preserved | Preserved |
| Context awareness | Not preserved | Preserved |

## What is Semantic Search (Dot Product and Cosine Similarity)?

**Semantic Search** = Meaning-based intelligent search using embeddings.

Semantic search means searching based on meaning, not exact words.

Semantic search is usually performed using:

- Cosine Similarity
- Dot Product
- Euclidean Distance

### Example

- Eating fiber reduce heart risk.
- Eating fruits and vegetable lowers cardiovascular disease chances.

Both sentences have the same meaning.

Even though the words are different, embeddings place them close together in vector space.

## Why Semantic Search?

Semantic search returns relevant results even when wording changes.

**Examples:**
- Eating fiber reduces heart risk.
- Buying a new car improves driving comfort.

### Complexity: Different Topics
- Health vs. Automobile
- Embeddings will be far apart due to low cosine similarity.
